```python
df.columns=df.columns.str.lower()
```

```python
df.columns=df.columns.str.replace(" ","_")
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df["credit_score"]=df["credit_score"].fillna(df["credit_score"].median())
```

```python
df["annual_income"]=df["annual_income"].fillna(df["annual_income"].median())
```

```python
df["years_in_current_job"]=df["years_in_current_job"].fillna(df["years_in_current_job"].mode()[0])
```

```python
df["home_ownership"]=df["home_ownership"].fillna(df["home_ownership"].mode()[0])
```

```python
df["purpose"]=df["purpose"].fillna(df["purpose"].mode()[0])
```

```python
df["monthly_debt"]=df["monthly_debt"].fillna(df["monthly_debt"].median())
```

```python
df["current_loan_amount"]=df["current_loan_amount"].fillna(df["current_loan_amount"].median())
```

```python
df["term"]=df["term"].fillna(df["term"].mode()[0])
```

```python
df["years_of_credit_history"]=df["years_of_credit_history"].fillna(df["years_of_credit_history"].median())
```

```python
df["months_since_last_delinquent"]=df["months_since_last_delinquent"].fillna(df["months_since_last_delinquent"].median())
```

```python
df["number_of_open_accounts"]=df["number_of_open_accounts"].fillna(df["number_of_open_accounts"].median())
```

```python
df["number_of_credit_problems"]=df["number_of_credit_problems"].fillna(0)
```

```python
df["maximum_open_credit"]=df["maximum_open_credit"].fillna(df["maximum_open_credit"].median())
```

```python
df["current_credit_balance"]=df["current_credit_balance"].fillna(df["current_credit_balance"].median())
```

```python
df["bankruptcies"]=df["bankruptcies"].fillna(0)
```

```python
df["tax_liens"]=df["tax_liens"].fillna(0)
```

```python
# new column credit category
df["credit_category"]=pd.cut(
    df["credit_score"],
    bins=[0,580,670,740,800,1000],
    labels=["Poor","Fair","Good","Very Good","Excellent"]
)
```

```python
df.isnull().sum()
```

```python
df["credit_category"]=df["credit_category"].fillna(df["credit_category"].mode()[0])
```

```python
df["credit_category"]
```

```python
#debt to income ratio
df["debt_to_income"]=df["monthly_debt"]/df["annual_income"]
```

```python
df["debt_to_income"]
```

```python
#High risk customer
df["risk_level"]=df["credit_score"].apply(
    lambda x:"High Risk" if x<670 else "Low")
```

```python
df["risk_level"]
```

```python
#average credit score by home analysis
df.groupby("home_ownership")["credit_score"].mean()
```

```python
#loan amount by purpose
df.groupby("purpose")["current_loan_amount"].mean()
```

```python
#top 10 largest loans by purpose
df.nlargest(10,"current_loan_amount")[["current_loan_amount","purpose"]]
```
